```python
# Select credit limit data
data = df['LIMIT_BAL'].dropna()

# Calculate mean and standard deviation
mu = data.mean()
sigma = data.std()

print("Mean (mu):", mu)
print("Standard Deviation (sigma):", sigma)
print("Skewness:", data.skew())
print("Kurtosis:", data.kurtosis())
```

```python
data = df['LIMIT_BAL'].dropna()

mu = data.mean()
sigma = data.std()

plt.figure(figsize=(10, 6))

sns.histplot(
    data,
    bins=30,
    stat='density',
    color='skyblue',
    label='Actual Credit Limit'
)

x = np.linspace(data.min(), data.max(), 1000)

plt.plot(
    x,
    norm.pdf(x, mu, sigma),
    color='red',
    linewidth=2,
    label='Normal Distribution'
)

plt.title('LIMIT_BAL vs Normal Distribution')
plt.xlabel('Credit Limit (NTD)')
plt.ylabel('Density')
plt.legend()
plt.show()
    
```

3. Улаан муруй бодит тархалттай бүрэн тохирохгүй байна. Иймээс LIMIT_BAL нь хэвийн тархалтаас мэдэгдэхүйц ялгаатай харагдаж байна.

   Normal Distribution Analysis of LIMIT_BAL

The histogram shows the actual distribution of credit limits among clients, while the red curve represents a theoretical normal distribution based on the sample mean and standard deviation.

The actual distribution is positively skewed, with most observations concentrated at lower credit limits and a long tail extending towards higher values. The normal curve does not closely match the histogram, indicating that the credit limit data may not follow a normal distribution.

Therefore, the visual comparison suggests that LIMIT_BAL is not normally distributed. This conclusion can be further supported by examining the skewness value, a Q-Q plot, and a normality test.

```python
print("Skewness:", df['LIMIT_BAL'].skew())
print("Kurtosis:", df['LIMIT_BAL'].kurtosis())
```

Skewness > 0: Баруун тийш хазайсан.
Skewness ≈ 0: Тэгш хэмтэй байж болзошгүй.
Skewness < 0: Зүүн тийш хазайсан.

```python
from scipy.stats import binom, poisson
# Binomial distribution
n = 20
k = np.arange(0, n + 1)

binom_probs = binom.pmf(k, n, p)

print("P(X = 3):", round(binom.pmf(3, n, p), 4))
print("Percentage:", round(binom.pmf(3, n, p) * 100, 2), "%")

# Plot
plt.figure(figsize=(9, 5))
plt.bar(k, binom_probs)
plt.title('Binomial Distribution of Credit Card Defaults')
plt.xlabel('Number of Defaults (k)')
plt.ylabel('Probability')
plt.xticks(k)
plt.show()
```

```python

# Poisson distribution
lam = n * p
poisson_probs = poisson.pmf(k, lam)

print("Lambda:", round(lam, 4))
print("P(X = 3):", round(poisson.pmf(3, lam), 4))

plt.figure(figsize=(9, 5))
plt.bar(k, poisson_probs)
plt.title('Poisson Distribution of Credit Card Defaults')
plt.xlabel('Number of Defaults (k)')
plt.ylabel('Probability')
plt.xticks(k)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson

# Estimate default probability from the dataset
p = df['default payment next month'].mean()

# Set parameters
n = 20
lam = n * p
k = np.arange(0, n + 1)

# Calculate probabilities
binom_probs = binom.pmf(k, n, p)
poisson_probs = poisson.pmf(k, lam)

# Compare Binomial and Poisson distributions
plt.figure(figsize=(10, 6))

plt.plot(
    k, binom_probs,
    marker='o',
    label='Binomial Distribution'
)

plt.plot(
    k, poisson_probs,
    marker='x',
    linestyle='--',
    label='Poisson Distribution'
)

plt.title('Binomial vs Poisson Distribution of Credit Card Defaults')
plt.xlabel('Number of Defaults (k)')
plt.ylabel('Probability')
plt.xticks(k)
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# Compare probabilities numerically
print("Default probability (p):", round(p, 4))
print("Number of clients (n):", n)
print("Poisson lambda (n × p):", round(lam, 4))
print("Binomial P(X = 3):", round(binom.pmf(3, n, p), 4))
print("Poisson P(X = 3):", round(poisson.pmf(3, lam), 4))
```
